```python
MODEL_COLS = [
    "WizardLM/WizardLM-13B-V1.2",
    "claude-instant-v1",
    "claude-v1",
    "claude-v2",
    "gpt-3.5-turbo-1106",
    "gpt-4-1106-preview",
    "meta/code-llama-instruct-34b-chat",
    "meta/llama-2-70b-chat",
    "mistralai/mistral-7b-chat",
    "mistralai/mixtral-8x7b-chat",
    "zero-one-ai/Yi-34B-Chat"
]

print("Number of models:", len(MODEL_COLS))

for model in MODEL_COLS:
    print(model)
```

```python
mask_mmlu = df["eval_name"].str.startswith("mmlu-", na=False)
mask_arc = df["eval_name"].eq("arc-challenge")
mask_gsm8k = df["eval_name"].eq("grade-school-math")

project_df = df[
    mask_mmlu | mask_arc | mask_gsm8k
].copy()

print("Original dataset:", len(df))
print("Project subset:", len(project_df))

print("\nRaw task counts:")
print(project_df["eval_name"].value_counts())
```

```python
import numpy as np

project_df["domain"] = np.select(
    [
        project_df["eval_name"].str.startswith("mmlu-"),
        project_df["eval_name"].eq("arc-challenge"),
        project_df["eval_name"].eq("grade-school-math")
    ],
    [
        "MMLU",
        "ARC",
        "GSM8K"
    ],
    default="OTHER"
)

print("\nFinal domain counts:")
print(project_df["domain"].value_counts())

print("\nTotal:")
print(len(project_df))
```

```python
score_df = project_df[MODEL_COLS].apply(
    pd.to_numeric,
    errors="coerce"
)

fractional_mask = (
    (score_df > 0) &
    (score_df < 1)
)

print(
    "Total model-question scores:",
    score_df.size
)

print(
    "Fractional scores:",
    fractional_mask.sum().sum()
)

print(
    "Questions containing at least one fractional score:",
    fractional_mask.any(axis=1).sum()
)
```

```python
for domain in ["MMLU", "ARC", "GSM8K"]:

    idx = project_df["domain"] == domain

    sub_scores = score_df.loc[idx]

    fractional = (
        (sub_scores > 0) &
        (sub_scores < 1)
    )

    print("\n", domain)

    print(
        "Questions:",
        len(sub_scores)
    )

    print(
        "Questions with fractional scores:",
        fractional.any(axis=1).sum()
    )

    print(
        "Total fractional cells:",
        fractional.sum().sum()
    )
```

```python
# Rebuild everything needed for fractional-score inspection

score_df = project_df[MODEL_COLS].apply(
    pd.to_numeric,
    errors="coerce"
)

fractional_mask = (
    (score_df > 0) &
    (score_df < 1)
)

fractional_rows = fractional_mask.any(axis=1)

print(
    "Questions containing at least one fractional score:",
    fractional_rows.sum()
)

# Inspect the first fractional example
row = project_df.loc[fractional_rows].iloc[0]

print("\nSample ID:")
print(row["sample_id"])

print("\nDomain:")
print(row["domain"])

print("\nTask:")
print(row["eval_name"])

print("\nPrompt:")
print(row["prompt"])

print("\nModel scores:")
for model in MODEL_COLS:
    print(model, "->", row[model])

print("\nOracle:")
print(row["oracle_model_to_route_to"])
```
